Complete the function below. <br>
The plot_featuremap function takes a set of feature maps and creates a grid of subplots, each displaying a single feature map. 

````python
### START CODE HERE ###
def plot_featuremap(img, title):
    if torch.is_tensor(img):
        img = img.detach().numpy()
    
    if img.ndim == 4:
        img = img.squeeze(0)
    
    num_channels = img.shape[0]
    cols = 8
    rows = (num_channels + cols - 1) // cols
    
    plt.figure(figsize=(16, rows * 2))
    
    for i in range(num_channels):
        plt.subplot(rows, cols, i + 1)
        feature_map = img[i]
        plt.imshow(feature_map, cmap='gray')
        plt.title(f'Channel {i}')
        plt.axis('off')
    
    plt.suptitle(title, fontsize=16)
    plt.tight_layout()
    plt.show()
### END CODE HERE ###
````

## Convolution 2D from scratch
Complete the function below

````python
### START CODE HERE ###
def convolution2d(img, kernel, padding, stride):
    if img.ndim == 3:
        H, W, C = img.shape
    else:
        H, W = img.shape
        C = 1
        img = img.reshape(H, W, 1)
    
    kH, kW = kernel.shape[:2]
    out_H = (H + 2 * padding - kH) // stride + 1
    out_W = (W + 2 * padding - kW) // stride + 1

    if padding > 0:
        img_padded = np.zeros((H + 2*padding, W + 2*padding, C))
        img_padded[padding:padding+H, padding:padding+W, :] = img
    else:
        img_padded = img

    if kernel.ndim == 3:
        output = np.zeros((out_H, out_W))
    else:
        output = np.zeros((out_H, out_W))

    for i in range(0, out_H):
        for j in range(0, out_W):
            h_start = i * stride
            h_end = h_start + kH
            w_start = j * stride
            w_end = w_start + kW
            
            region = img_padded[h_start:h_end, w_start:w_end, :]
            
            if kernel.ndim == 3:
                output[i, j] = np.sum(region * kernel)
            else:
                output[i, j] = np.sum(region[:, :, 0] * kernel)
    
    return output

### END CODE HERE ###
````

**ตรวจจับสี (color-sensitive kernels):**
- หาก `kernel` มีค่า `weight` ที่ต่างกันชัดเจนระหว่าง `channel` (เช่น `channel R` มีค่าสูงบวก ขณะที่ `G/B` มีค่าน้อยหรือลบ) `kernel` นั้นจะตอบสนองเมื่อมีส่วนของภาพที่เน้นสีที่สอดคล้องกับเชิง `weight` — นั่นคือมันทำหน้าที่เป็นตัวกรองสีเชิงเฉพาะช่อง `(color filter)`
- ถ้าใน `kernel` ค่าเชิงรวมบนช่อง `R` เป็นบวกสูงกว่า `G/B` → จะเน้นโครงสร้างที่เด่นในช่องแดง

**ตรวจจับขอบ/รายละเอียดเชิงโครงสร้าง (edge-sensitive kernels):**
- ถ้า `kernel` มีรูปแบบ `spatial` ที่มีค่าบวกในตำแหน่งหนึ่งและลบในตำแหน่งข้างเคียง จะจับความต่างเชิงพื้นที่ `(gradient)` → ตรวจจับขอบได้ดี
- ข้อสำคัญคือ `multi-channel pattern` และการสลับเครื่องหมาย (+/-) ภายใน `kernel` ทำให้ตอบสนองต่อการเปลี่ยนแปลงเชิงพื้นที่มากกว่าค่าค่าสีเฉพาะ

<br>**2.** <br>
`Matplotlib/plt.imshow` สามารถแสดงค่าลบเพราะมันไม่จำเป็นต้องมีค่าบวกเพื่อแสดงภาพ
- `imshow` จะแมปค่าจากช่วงข้อมูล `(data range)` ไปยัง `colormap` โดยอัตโนมัติ (หรือเมื่อกำหนด `vmin`/`vmax`/`Normalize`) — ดังนั้นค่าลบก็จะถูก `scale` ไปยังระดับสีที่เหมาะสม (เช่น ค่าต่ำเป็นสีเข้ม ค่าสูงเป็นสีอ่อน)
- ถ้าต้องการมองให้เป็นภาพแบบภาพจริง ก็สามารถ `normalize` หรือ `shift+scale` ค่า (เช่น `add offset` แล้ว `clip`) ก่อนแสดงผล

<br>**3.** 
- `im2col + GEMM (flatten patches → matrix multiply)`: แปลง `convolution` เป็น `matrix-matrix multiply` ใช้ `BLAS`/`optimized GEMM` เร็วมากบน `CPU`/`GPU`
- `FFT-based convolution`: ใช้ `convolution theorem` `(O(n log n))` เหมาะกับ `kernel` ใหญ่และภาพใหญ่
- `Winograd minimal filtering`: ลดจำนวน `multiplications` สำหรับ `kernel` ขนาดเล็ก (เช่น 3x3)
- `Separable convolution`: ถ้าเคอร์เนลเป็น `separable` `(k(x,y)=a(x)*b(y))` ให้ทำ `convolution` ตามแกนทีละแกน `(O(n) ลดลง)`
